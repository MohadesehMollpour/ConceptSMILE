```python
import torch
import numpy as np

messages = [
    {
        "role": "user",
        "content": [
            {"type": "image", "image": image},
            {
                "type": "text",
                "text": """
You are an expert retinal image analyst.

Analyze this retinal image and evaluate only these three concepts:

1. blood_vessels = visible retinal blood vessels
2. lesion = any visible retinal abnormality or pathological sign, including hemorrhage, exudate, microaneurysm, edema, or other anomaly
3. optic_disc = visible optic disc

Decision rule:
- Answer "yes" only if the concept is clearly visible in the image.
- Answer "no" if the concept is not visible or if you are uncertain.

Return only one valid JSON object.
Do not use markdown.
Do not add explanation.
Do not add extra text.
Do not include any concept other than these three.

Output format:
{
  "blood_vessels": "yes",
  "lesion": "yes",
  "optic_disc": "yes"
}
"""
            }
        ]
    }
]

text_prompt = processor.apply_chat_template(
    messages,
    tokenize=False,
    add_generation_prompt=True
)

inputs = processor(
    text=[text_prompt],
    images=[image],
    padding=True,
    return_tensors="pt"
)

inputs = {k: v.to(model.device) if hasattr(v, "to") else v for k, v in inputs.items()}

with torch.no_grad():
    generation_output = model.generate(
        **inputs,
        max_new_tokens=60,
        do_sample=False,
        return_dict_in_generate=True,
        output_scores=True
    )

generated_ids = generation_output.sequences[:, inputs["input_ids"].shape[1]:]

response = processor.batch_decode(
    generated_ids,
    skip_special_tokens=True,
    clean_up_tokenization_spaces=True
)[0]

print("Model response:")
print(response)

transition_scores = model.compute_transition_scores(
    generation_output.sequences,
    generation_output.scores,
    normalize_logits=True
)

generated_token_ids = generated_ids[0]
generated_tokens = processor.tokenizer.convert_ids_to_tokens(generated_token_ids)

token_score_data = []
for tok, score in zip(generated_tokens, transition_scores[0][-len(generated_tokens):]):
    token_score_data.append({
        "token": tok,
        "logprob_like_score": float(score.cpu().numpy()),
        "probability_like": float(np.exp(score.cpu().numpy()))
    })

print("\nToken scores:")
for row in token_score_data:
    print(row)
```

```python
import pandas as pd

token_df = pd.DataFrame(token_score_data)
display(token_df)

def extract_node_confidence_from_structured_json(token_score_data):
    """
    Expected JSON field order:
    1. blood_vessels
    2. lesion
    3. optic_disc

    Confidence is taken from the yes/no token associated with each field.
    If answer == "yes", confidence = P(yes)
    If answer == "no", confidence = 1 - P(no)
    """
    concept_order = ["blood_vessels", "lesion", "optic_disc"]
    field_token_map = {
        "blood_vessels": ["blood", "_v", "ess", "els"],
        "lesion": ["les", "ion"],
        "optic_disc": ["optic", "_disc"],
    }

    results = {}
    i = 0
    current_concept_idx = 0

    while i < len(token_score_data) and current_concept_idx < len(concept_order):
        concept_name = concept_order[current_concept_idx]
        field_tokens = field_token_map[concept_name]

        match = True
        if i + len(field_tokens) <= len(token_score_data):
            for j, piece in enumerate(field_tokens):
                if token_score_data[i + j]["token"] != piece:
                    match = False
                    break
        else:
            match = False

        if match:
            k = i + len(field_tokens)
            answer_token = None
            confidence = None

            while k < len(token_score_data):
                t = token_score_data[k]["token"].replace("Ġ", "").replace('"', "").strip().lower()

                if t in ["yes", "no"]:
                    answer_token = t
                    prob = float(token_score_data[k]["probability_like"])
                    confidence = prob if t == "yes" else (1.0 - prob)
                    break
                k += 1

            results[concept_name] = {
                "answer": answer_token,
                "confidence": confidence
            }

            current_concept_idx += 1
            i = k if k > i else i + 1
        else:
            i += 1

    return results

node_confidence = extract_node_confidence_from_structured_json(token_score_data)

print("Node confidence results:")
for concept, info in node_confidence.items():
    print(f"{concept}: answer={info['answer']}, confidence={info['confidence']:.6f}")

node_confidence_df = pd.DataFrame([
    {
        "node_name": concept,
        "answer": info["answer"],
        "confidence": info["confidence"]
    }
    for concept, info in node_confidence.items()
])

print("\nStructured node confidence table:")
display(node_confidence_df)
```

```python
import numpy as np
import pandas as pd
import torch
from PIL import Image

if "IMG_NP" not in globals():
    raise ValueError("IMG_NP not found. Run the superpixel/image cell first.")

if "perturbed_images" not in globals():
    raise ValueError("perturbed_images not found. Run the perturbation cell first.")

if "processor_dino" not in globals() or "model_dino" not in globals():
    raise ValueError("DINO model not loaded. Run the DINO loading cell first.")

if "device_dino" not in globals():
    raise ValueError("device_dino not found. Run the DINO loading cell first.")

def get_dino_embedding(img_np):
    if img_np.dtype != np.uint8:
        img_np = img_np.astype(np.uint8)

    pil_img = Image.fromarray(img_np).convert("RGB")
    inputs = processor_dino(images=pil_img, return_tensors="pt")
    inputs = {k: v.to(device_dino) for k, v in inputs.items()}

    with torch.no_grad():
        outputs = model_dino(**inputs)

    # CLS token embedding
    emb = outputs.last_hidden_state[:, 0, :]
    emb = emb.squeeze(0).detach().cpu().numpy().astype(np.float32)
    return emb

def cosine_distance(a, b):
    a = np.asarray(a, dtype=np.float32)
    b = np.asarray(b, dtype=np.float32)

    denom = (np.linalg.norm(a) * np.linalg.norm(b)) + 1e-12
    cos_sim = float(np.dot(a, b) / denom)
    cos_dist = 1.0 - cos_sim
    return cos_sim, cos_dist

print("Extracting original image embedding...")
orig_emb = get_dino_embedding(IMG_NP)

rows = []
sigma = 0.25

for i, pert_img in enumerate(perturbed_images):
    pert_emb = get_dino_embedding(pert_img)
    cos_sim, cos_dist = cosine_distance(orig_emb, pert_emb)

    kernel_weight = float(np.exp(- (cos_dist ** 2) / (sigma ** 2)))

    rows.append({
        "perturb_id": i,
        "cosine_similarity": cos_sim,
        "cosine_distance": cos_dist,
        "kernel_weight_from_cosdist": kernel_weight
    })

distance_df = pd.DataFrame(rows)

print("distance_df created")
display(distance_df.head())

distance_df.to_csv("/kaggle/working/dino_cosine_distance_weights.csv", index=False)
print("Saved: /kaggle/working/dino_cosine_distance_weights.csv")
```

```python
# ============================================================
# NEW CELL AFTER YOUR CURRENT FIDELITY CELL
# Complete node-level fidelity table + variance
# Adapted to current Qwen pipeline
# ============================================================

import numpy as np
import pandas as pd

from xgboost import XGBRegressor
from sklearn.metrics import r2_score, mean_squared_error, mean_absolute_error
from sklearn.model_selection import train_test_split

# -----------------------------
# Safety checks
# -----------------------------
if "perturb_vectors" not in globals():
    raise ValueError("perturb_vectors is missing.")

if "node_compare_df" not in globals():
    raise ValueError("node_compare_df is missing. Run the node comparison cell first.")

if "distance_df" not in globals():
    raise ValueError("distance_df is missing. Run the cosine distance cell first.")

X_all = np.array(perturb_vectors, dtype=float)

if X_all.ndim != 2:
    raise ValueError("perturb_vectors must be a 2D array.")

if len(node_compare_df) == 0:
    raise ValueError("node_compare_df is empty.")

# -----------------------------
# Build node_shift_matrix from node_compare_df
# target = conf_shift
# -----------------------------
node_shift_matrix = node_compare_df.pivot_table(
    index="perturb_id",
    columns="node_name",
    values="conf_shift",
    aggfunc="first"
).sort_index()

if len(node_shift_matrix) == 0:
    raise ValueError("node_shift_matrix is empty after pivot.")

# -----------------------------
# Align perturb_vectors with node_shift_matrix
# -----------------------------
valid_ids = node_shift_matrix.index.to_numpy().astype(int)

if valid_ids.max() >= len(X_all):
    raise ValueError("Some perturb_id values exceed perturb_vectors length.")

X_all = X_all[valid_ids]

# -----------------------------
# Weight utilities
# -----------------------------
def weighted_mean(x, w):
    x = np.asarray(x, dtype=float)
    w = np.asarray(w, dtype=float)
    w_sum = np.sum(w)

    if len(x) == 0:
        return np.nan
    if w_sum <= 0:
        return np.mean(x)

    return np.sum(w * x) / w_sum

def weighted_mse(y_true, y_pred, w):
    return weighted_mean((np.asarray(y_true) - np.asarray(y_pred)) ** 2, w)

def weighted_mae(y_true, y_pred, w):
    return weighted_mean(np.abs(np.asarray(y_true) - np.asarray(y_pred)), w)

def weighted_r2(y_true, y_pred, w):
    y_true = np.asarray(y_true, dtype=float)
    y_pred = np.asarray(y_pred, dtype=float)
    w = np.asarray(w, dtype=float)

    if len(y_true) < 2:
        return np.nan

    y_bar = weighted_mean(y_true, w)
    ss_res = np.sum(w * (y_true - y_pred) ** 2)
    ss_tot = np.sum(w * (y_true - y_bar) ** 2)

    if ss_tot <= 1e-12:
        return np.nan

    return 1.0 - (ss_res / ss_tot)

def kish_effective_n(w):
    w = np.asarray(w, dtype=float)
    denom = np.sum(w ** 2)

    if denom <= 0:
        return len(w)

    return (np.sum(w) ** 2) / denom

def weighted_adjusted_r2(y_true, y_pred, w, p):
    r2w = weighted_r2(y_true, y_pred, w)

    if pd.isna(r2w):
        return np.nan

    n_eff = kish_effective_n(w)

    if n_eff <= (p + 1):
        return np.nan

    return 1.0 - (1.0 - r2w) * ((n_eff - 1.0) / (n_eff - p - 1.0))

# -----------------------------
# Build sample weights from distance_df
# -----------------------------
def get_weight_map():
    weight_map = pd.Series(np.ones(len(valid_ids), dtype=float), index=valid_ids)

    if "perturb_id" not in distance_df.columns:
        print("distance_df has no perturb_id -> using equal weights.")
        return weight_map, "equal_weights"

    if "kernel_weight_from_cosdist" in distance_df.columns:
        wcol = "kernel_weight_from_cosdist"
    elif "kernel_weight" in distance_df.columns:
        wcol = "kernel_weight"
    else:
        print("No usable weight column in distance_df -> using equal weights.")
        return weight_map, "equal_weights"

    tmp = distance_df[["perturb_id", wcol]].copy()
    tmp[wcol] = pd.to_numeric(tmp[wcol], errors="coerce").fillna(1.0)
    tmp = tmp.drop_duplicates(subset=["perturb_id"]).set_index("perturb_id")[wcol]

    tmp = tmp.reindex(valid_ids).fillna(1.0)

    return tmp, wcol

# -----------------------------
# Main repeated-run fidelity
# -----------------------------
N_REPEATS = 30
TEST_SIZE = 0.30

weight_map, weight_source = get_weight_map()

all_runs = []

for node in node_shift_matrix.columns:
    y_full = node_shift_matrix[node].values.astype(float)

    valid_mask = ~np.isnan(y_full)
    X = X_all[valid_mask]
    y = y_full[valid_mask]

    node_valid_ids = node_shift_matrix.index.to_numpy()[valid_mask]
    w = weight_map.reindex(node_valid_ids).fillna(1.0).values.astype(float)

    if len(y) < 8:
        all_runs.append({
            "node_name": node,
            "repeat_id": np.nan,
            "status": "too_few_samples",
            "MSE": np.nan,
            "MAE": np.nan,
            "Lm": np.nan,
            "mean_L1": np.nan,
            "mean_L2": np.nan,
            "WMSE": np.nan,
            "WMAE": np.nan,
            "weighted_R2": np.nan,
            "weighted_adjusted_R2": np.nan,
            "R2": np.nan
        })
        continue

    if np.std(y) < 1e-8:
        for rep in range(N_REPEATS):
            all_runs.append({
                "node_name": node,
                "repeat_id": rep,
                "status": "no_variation",
                "MSE": 0.0,
                "MAE": 0.0,
                "Lm": 0.0,
                "mean_L1": 0.0,
                "mean_L2": 0.0,
                "WMSE": 0.0,
                "WMAE": 0.0,
                "weighted_R2": np.nan,
                "weighted_adjusted_R2": np.nan,
                "R2": np.nan
            })
        continue

    for rep in range(N_REPEATS):
        X_train, X_test, y_train, y_test, w_train, w_test = train_test_split(
            X, y, w,
            test_size=TEST_SIZE,
            random_state=42 + rep
        )

        model = XGBRegressor(
            n_estimators=200,
            max_depth=3,
            learning_rate=0.05,
            subsample=0.8,
            colsample_bytree=0.8,
            reg_lambda=1.0,
            random_state=42 + rep
        )

        model.fit(X_train, y_train, sample_weight=w_train)
        y_pred = model.predict(X_test)

        mse_val = mean_squared_error(y_test, y_pred)
        mae_val = mean_absolute_error(y_test, y_pred)
        mean_l1_val = np.mean(np.abs(y_test - y_pred))
        mean_l2_val = np.mean((y_test - y_pred) ** 2)
        lm_val = 0.5 * (mean_l1_val + mean_l2_val)

        wmse_val = weighted_mse(y_test, y_pred, w_test)
        wmae_val = weighted_mae(y_test, y_pred, w_test)
        wr2_val = weighted_r2(y_test, y_pred, w_test)
        wr2_adj_val = weighted_adjusted_r2(
            y_true=y_test,
            y_pred=y_pred,
            w=w_test,
            p=X_test.shape[1]
        )

        all_runs.append({
            "node_name": node,
            "repeat_id": rep,
            "status": "ok",
            "MSE": mse_val,
            "MAE": mae_val,
            "Lm": lm_val,
            "mean_L1": mean_l1_val,
            "mean_L2": mean_l2_val,
            "WMSE": wmse_val,
            "WMAE": wmae_val,
            "weighted_R2": wr2_val,
            "weighted_adjusted_R2": wr2_adj_val,
            "R2": r2_score(y_test, y_pred)
        })

node_fidelity_runs_df = pd.DataFrame(all_runs)

# -----------------------------
# Mean table per node
# -----------------------------
node_fidelity_complete_table = (
    node_fidelity_runs_df
    .groupby("node_name", as_index=False)
    .agg({
        "MSE": "mean",
        "MAE": "mean",
        "Lm": "mean",
        "mean_L1": "mean",
        "mean_L2": "mean",
        "WMSE": "mean",
        "WMAE": "mean",
        "R2": "mean",
        "weighted_R2": "mean",
        "weighted_adjusted_R2": "mean"
    })
)

# -----------------------------
# Variance table per node
# -----------------------------
node_fidelity_variance_table = (
    node_fidelity_runs_df
    .groupby("node_name", as_index=False)
    .agg({
        "R2": ["var", "std"],
        "weighted_R2": ["var", "std"]
    })
)

node_fidelity_variance_table.columns = [
    "node_name",
    "var_R2", "std_R2",
    "var_weighted_R2", "std_weighted_R2"
]

# -----------------------------
# Final merged table
# -----------------------------
node_fidelity_complete_table = node_fidelity_complete_table.merge(
    node_fidelity_variance_table,
    on="node_name",
    how="left"
).sort_values("weighted_R2", ascending=False)

print("Complete node-level fidelity table")
print("Weight source:", weight_source)
display(node_fidelity_complete_table.round(4))

# save
node_fidelity_runs_df.to_csv("/kaggle/working/node_fidelity_runs_df.csv", index=False)
node_fidelity_complete_table.to_csv("/kaggle/working/node_fidelity_complete_table.csv", index=False)

print("Saved:")
print("/kaggle/working/node_fidelity_runs_df.csv")
print("/kaggle/working/node_fidelity_complete_table.csv")
```

```python
# ============================================================
# NEW CELL: Thesis-style concept-level fidelity table
# Adapted for current Qwen pipeline
# ============================================================

import numpy as np
import pandas as pd

# -----------------------------
# Safety check
# -----------------------------
if "node_fidelity_complete_table" not in globals():
    raise ValueError(
        "node_fidelity_complete_table is missing. Run the complete node-level fidelity cell first."
    )

df = node_fidelity_complete_table.copy()

if "node_name" not in df.columns:
    raise ValueError("node_fidelity_complete_table must contain 'node_name' column.")

# -----------------------------
# Map each node to thesis concept
# -----------------------------
def map_concept(node_name):
    s = str(node_name).lower().strip().replace("-", "_").replace(" ", "_")

    if "lesion" in s:
        return "lesion"
    elif "vessel" in s or "blood_vessel" in s:
        return "vessel"
    elif "optic" in s and ("disc" in s or "disk" in s):
        return "optic_disc"
    else:
        return np.nan

df["concept"] = df["node_name"].apply(map_concept)
df = df.dropna(subset=["concept"]).copy()

if df.empty:
    raise ValueError("No lesion / vessel / optic_disc nodes found.")

# -----------------------------
# Ensure all required columns exist (important fix)
# -----------------------------
required_cols = [
    "MSE","MAE","Lm","mean_L1","mean_L2",
    "WMSE","WMAE","R2","weighted_R2","weighted_adjusted_R2",
    "var_R2","std_R2","var_weighted_R2","std_weighted_R2"
]

for col in required_cols:
    if col not in df.columns:
        df[col] = np.nan

# -----------------------------
# Aggregate by concept
# -----------------------------
concept_level_fidelity_table = (
    df.groupby("concept", as_index=False)
      .agg({
          "MSE": "mean",
          "MAE": "mean",
          "Lm": "mean",
          "mean_L1": "mean",
          "mean_L2": "mean",
          "WMSE": "mean",
          "WMAE": "mean",
          "R2": "mean",
          "weighted_R2": "mean",
          "weighted_adjusted_R2": "mean",
          "var_R2": "mean",
          "std_R2": "mean",
          "var_weighted_R2": "mean",
          "std_weighted_R2": "mean"
      })
)

# -----------------------------
# Rename for thesis
# -----------------------------
concept_level_fidelity_table = concept_level_fidelity_table.rename(columns={
    "concept": "Concept",
    "MSE": "MSE",
    "MAE": "MAE",
    "Lm": "Lm",
    "mean_L1": "mean-L1",
    "mean_L2": "mean-L2",
    "WMSE": "WMSE",
    "WMAE": "WMAE",
    "R2": "R2",
    "weighted_R2": "R2ω",
    "weighted_adjusted_R2": "R̂2ω",
    "var_R2": "Var(R2)",
    "std_R2": "Std(R2)",
    "var_weighted_R2": "Var(R2ω)",
    "std_weighted_R2": "Std(R2ω)"
})

# -----------------------------
# Ordering
# -----------------------------
order_map = {"lesion": 0, "vessel": 1, "optic_disc": 2}

concept_level_fidelity_table["__order"] = concept_level_fidelity_table["Concept"].map(order_map)
concept_level_fidelity_table = concept_level_fidelity_table.sort_values("__order").drop(columns="__order")

print("Thesis-style concept-level fidelity table:")
display(concept_level_fidelity_table.round(4))

# Save
concept_level_fidelity_table.to_csv("/kaggle/working/concept_level_fidelity_table.csv", index=False)

print("Saved:")
print("/kaggle/working/concept_level_fidelity_table.csv")
```

```python
# ============================================================
# CORRECT VLM ATT (Cosine-weighted version)
# Concept-level ATT metrics
# Based on confidence change (conf_shift)
# ============================================================

import numpy as np
import pandas as pd
from sklearn.metrics import roc_auc_score, roc_curve, accuracy_score, f1_score

# -----------------------------
# Checks
# -----------------------------
if "node_compare_df" not in globals():
    raise ValueError("node_compare_df is missing.")

if "distance_df" not in globals():
    raise ValueError("distance_df (cosine) is missing.")

# -----------------------------
# ATT score = |conf_shift|
# -----------------------------
att_df = node_compare_df.copy()
att_df["att_score"] = att_df["conf_shift"].abs()

# -----------------------------
# 🔥 CORRECT y_true (KEY FIX)
# -----------------------------
# Adaptive threshold per node (robust)

y_true_all = []

for node_name, g in att_df.groupby("node_name"):
    shifts = g["conf_shift"].abs().values

    if len(shifts) == 0:
        continue

    threshold = np.percentile(shifts, 75)

    for val in shifts:
        y_true_all.append(int(val >= threshold))

att_df["y_true"] = y_true_all

# -----------------------------
# Add Cosine weights
# -----------------------------
w_df = distance_df.copy().set_index("perturb_id")

if "kernel_weight_from_cosdist" not in w_df.columns:
    raise ValueError("distance_df must contain kernel_weight_from_cosdist")

att_df["weight"] = att_df["perturb_id"].map(
    w_df["kernel_weight_from_cosdist"]
).fillna(1.0)

# -----------------------------
# Concept mapping
# -----------------------------
def map_concept(node_name):
    s = str(node_name).lower()
    if "lesion" in s:
        return "lesion"
    elif "vessel" in s:
        return "vessel"
    elif "optic" in s:
        return "optic_disc"
    else:
        return np.nan

att_df["concept"] = att_df["node_name"].apply(map_concept)
att_df = att_df.dropna(subset=["concept"])

# -----------------------------
# Compute metrics per concept
# -----------------------------
rows = []

for concept, g in att_df.groupby("concept"):

    if len(g) < 5 or g["y_true"].nunique() < 2:
        rows.append({
            "Concept": concept,
            "ATT ACC": np.nan,
            "ATT F1": np.nan,
            "ATT AUROC": np.nan,
            "N": len(g)
        })
        continue

    y_true = g["y_true"].values
    scores = g["att_score"].values
    weights = g["weight"].values

    try:
        auc = roc_auc_score(y_true, scores, sample_weight=weights)
    except:
        auc = np.nan

    try:
        fpr, tpr, thr = roc_curve(y_true, scores)
        j = tpr - fpr
        best_thr = thr[np.argmax(j)]

        y_pred = (scores >= best_thr).astype(int)

        acc = accuracy_score(y_true, y_pred)
        f1 = f1_score(y_true, y_pred, zero_division=0)
    except:
        acc = np.nan
        f1 = np.nan

    rows.append({
        "Concept": concept,
        "ATT ACC": acc,
        "ATT F1": f1,
        "ATT AUROC": auc,
        "N": len(g)
    })

att_table_cosine = pd.DataFrame(rows)

# -----------------------------
# Ordering
# -----------------------------
order_map = {"lesion": 0, "vessel": 1, "optic_disc": 2}
att_table_cosine["__order"] = att_table_cosine["Concept"].map(order_map)
att_table_cosine = att_table_cosine.sort_values("__order").drop(columns="__order")

print("✅ Correct Concept-level ATT Table (Cosine):")
display(att_table_cosine.round(4))

# Save
att_table_cosine.to_csv("/kaggle/working/att_table_cosine_correct.csv", index=False)
```

```python
# ============================================================
# NEW CELL AFTER YOUR CURRENT WASSERSTEIN FIDELITY CELL
# Complete node-level fidelity table + variance
# Wasserstein-weighted version
# Adapted for current Qwen pipeline
# ============================================================

import os
import numpy as np
import pandas as pd

from xgboost import XGBRegressor
from sklearn.metrics import r2_score, mean_squared_error, mean_absolute_error
from sklearn.model_selection import train_test_split

# -----------------------------
# Safety checks
# -----------------------------
if "perturb_vectors" not in globals():
    raise ValueError("perturb_vectors is missing.")

if "node_compare_df" not in globals():
    raise ValueError("node_compare_df is missing. Run the node comparison cell first.")

if "distance_df_wasserstein" not in globals():
    raise ValueError("distance_df_wasserstein is missing. Run the Wasserstein distance cell first.")

X_all_full = np.array(perturb_vectors, dtype=float)

if X_all_full.ndim != 2:
    raise ValueError("perturb_vectors must be a 2D array.")

if len(node_compare_df) == 0:
    raise ValueError("node_compare_df is empty.")

# -----------------------------
# Build node_shift_matrix from node_compare_df
# target = conf_shift
# -----------------------------
node_shift_matrix = node_compare_df.pivot_table(
    index="perturb_id",
    columns="node_name",
    values="conf_shift",
    aggfunc="first"
).sort_index()

if len(node_shift_matrix) == 0:
    raise ValueError("node_shift_matrix is empty after pivot.")

valid_ids = node_shift_matrix.index.to_numpy().astype(int)

if valid_ids.max() >= len(X_all_full):
    raise ValueError("Some perturb_id values exceed perturb_vectors length.")

X_all = X_all_full[valid_ids]

# -----------------------------
# Weight utilities
# -----------------------------
def weighted_mean(x, w):
    x = np.asarray(x, dtype=float)
    w = np.asarray(w, dtype=float)
    w_sum = np.sum(w)

    if len(x) == 0:
        return np.nan
    if w_sum <= 0:
        return np.mean(x)

    return np.sum(w * x) / w_sum

def weighted_mse(y_true, y_pred, w):
    return weighted_mean((np.asarray(y_true) - np.asarray(y_pred)) ** 2, w)

def weighted_mae(y_true, y_pred, w):
    return weighted_mean(np.abs(np.asarray(y_true) - np.asarray(y_pred)), w)

def weighted_r2(y_true, y_pred, w):
    y_true = np.asarray(y_true, dtype=float)
    y_pred = np.asarray(y_pred, dtype=float)
    w = np.asarray(w, dtype=float)

    if len(y_true) < 2:
        return np.nan

    y_bar = weighted_mean(y_true, w)
    ss_res = np.sum(w * (y_true - y_pred) ** 2)
    ss_tot = np.sum(w * (y_true - y_bar) ** 2)

    if ss_tot <= 1e-12:
        return np.nan

    return 1.0 - (ss_res / ss_tot)

def kish_effective_n(w):
    w = np.asarray(w, dtype=float)
    denom = np.sum(w ** 2)

    if denom <= 0:
        return len(w)

    return (np.sum(w) ** 2) / denom

def weighted_adjusted_r2(y_true, y_pred, w, p):
    r2w = weighted_r2(y_true, y_pred, w)

    if pd.isna(r2w):
        return np.nan

    n_eff = kish_effective_n(w)

    if n_eff <= (p + 1):
        return np.nan

    return 1.0 - (1.0 - r2w) * ((n_eff - 1.0) / (n_eff - p - 1.0))

# -----------------------------
# Build Wasserstein weight map
# -----------------------------
def get_weight_map(valid_ids):
    weight_map = pd.Series(np.ones(len(valid_ids), dtype=float), index=valid_ids)

    df_w = distance_df_wasserstein.copy()

    if not isinstance(df_w, pd.DataFrame) or df_w.empty:
        print("Wasserstein dataframe invalid/empty -> using equal weights.")
        return weight_map, "equal_weights"

    if "perturb_id" not in df_w.columns:
        print("Wasserstein dataframe has no perturb_id -> using equal weights.")
        return weight_map, "equal_weights"

    if "kernel_weight_from_wassdist" in df_w.columns:
        wcol = "kernel_weight_from_wassdist"
    elif "kernel_weight" in df_w.columns:
        wcol = "kernel_weight"
    else:
        print("No usable Wasserstein weight column found -> using equal weights.")
        return weight_map, "equal_weights"

    tmp = df_w[["perturb_id", wcol]].copy()
    tmp[wcol] = pd.to_numeric(tmp[wcol], errors="coerce").fillna(1.0)
    tmp = tmp.drop_duplicates(subset=["perturb_id"]).set_index("perturb_id")[wcol]
    tmp = tmp.reindex(valid_ids).fillna(1.0)

    return tmp, wcol

# -----------------------------
# Main repeated-run fidelity
# -----------------------------
N_REPEATS = 30
TEST_SIZE = 0.30

weight_map, weight_source = get_weight_map(valid_ids)

all_runs = []

for node in node_shift_matrix.columns:
    y_full = node_shift_matrix[node].values.astype(float)

    valid_mask = ~np.isnan(y_full)
    X = X_all[valid_mask]
    y = y_full[valid_mask]

    node_valid_ids = node_shift_matrix.index.to_numpy()[valid_mask]
    w = weight_map.reindex(node_valid_ids).fillna(1.0).values.astype(float)

    if len(y) < 8:
        all_runs.append({
            "node_name": node,
            "repeat_id": np.nan,
            "status": "too_few_samples",
            "MSE": np.nan,
            "MAE": np.nan,
            "Lm": np.nan,
            "mean_L1": np.nan,
            "mean_L2": np.nan,
            "WMSE": np.nan,
            "WMAE": np.nan,
            "weighted_R2_wass": np.nan,
            "weighted_adjusted_R2_wass": np.nan,
            "R2": np.nan
        })
        continue

    if np.std(y) < 1e-8:
        for rep in range(N_REPEATS):
            all_runs.append({
                "node_name": node,
                "repeat_id": rep,
                "status": "no_variation",
                "MSE": 0.0,
                "MAE": 0.0,
                "Lm": 0.0,
                "mean_L1": 0.0,
                "mean_L2": 0.0,
                "WMSE": 0.0,
                "WMAE": 0.0,
                "weighted_R2_wass": np.nan,
                "weighted_adjusted_R2_wass": np.nan,
                "R2": np.nan
            })
        continue

    for rep in range(N_REPEATS):
        X_train, X_test, y_train, y_test, w_train, w_test = train_test_split(
            X, y, w,
            test_size=TEST_SIZE,
            random_state=42 + rep
        )

        model = XGBRegressor(
            n_estimators=200,
            max_depth=3,
            learning_rate=0.05,
            subsample=0.8,
            colsample_bytree=0.8,
            reg_lambda=1.0,
            random_state=42 + rep
        )

        model.fit(X_train, y_train, sample_weight=w_train)
        y_pred = model.predict(X_test)

        mse_val = mean_squared_error(y_test, y_pred)
        mae_val = mean_absolute_error(y_test, y_pred)
        mean_l1_val = np.mean(np.abs(y_test - y_pred))
        mean_l2_val = np.mean((y_test - y_pred) ** 2)
        lm_val = 0.5 * (mean_l1_val + mean_l2_val)

        wmse_val = weighted_mse(y_test, y_pred, w_test)
        wmae_val = weighted_mae(y_test, y_pred, w_test)
        wr2_val = weighted_r2(y_test, y_pred, w_test)
        wr2_adj_val = weighted_adjusted_r2(
            y_true=y_test,
            y_pred=y_pred,
            w=w_test,
            p=X_test.shape[1]
        )

        all_runs.append({
            "node_name": node,
            "repeat_id": rep,
            "status": "ok",
            "MSE": mse_val,
            "MAE": mae_val,
            "Lm": lm_val,
            "mean_L1": mean_l1_val,
            "mean_L2": mean_l2_val,
            "WMSE": wmse_val,
            "WMAE": wmae_val,
            "weighted_R2_wass": wr2_val,
            "weighted_adjusted_R2_wass": wr2_adj_val,
            "R2": r2_score(y_test, y_pred)
        })

node_fidelity_runs_df_wass = pd.DataFrame(all_runs)

# -----------------------------
# Mean table per node
# -----------------------------
node_fidelity_complete_table_wass = (
    node_fidelity_runs_df_wass
    .groupby("node_name", as_index=False)
    .agg({
        "MSE": "mean",
        "MAE": "mean",
        "Lm": "mean",
        "mean_L1": "mean",
        "mean_L2": "mean",
        "WMSE": "mean",
        "WMAE": "mean",
        "R2": "mean",
        "weighted_R2_wass": "mean",
        "weighted_adjusted_R2_wass": "mean"
    })
)

# -----------------------------
# Variance table per node
# -----------------------------
node_fidelity_variance_table_wass = (
    node_fidelity_runs_df_wass
    .groupby("node_name", as_index=False)
    .agg({
        "R2": ["var", "std"],
        "weighted_R2_wass": ["var", "std"]
    })
)

node_fidelity_variance_table_wass.columns = [
    "node_name",
    "var_R2", "std_R2",
    "var_weighted_R2_wass", "std_weighted_R2_wass"
]

# -----------------------------
# Final merged table
# -----------------------------
node_fidelity_complete_table_wass = node_fidelity_complete_table_wass.merge(
    node_fidelity_variance_table_wass,
    on="node_name",
    how="left"
).sort_values("weighted_R2_wass", ascending=False)

print("Complete node-level fidelity table (Wasserstein-weighted)")
print("Weight source:", weight_source)
display(node_fidelity_complete_table_wass.round(4))

# -----------------------------
# Save as new files
# -----------------------------
node_fidelity_runs_df_wass.to_csv("/kaggle/working/node_fidelity_runs_df_wass.csv", index=False)
node_fidelity_complete_table_wass.to_csv("/kaggle/working/node_fidelity_complete_table_wass.csv", index=False)

print("Saved:")
print("/kaggle/working/node_fidelity_runs_df_wass.csv")
print("/kaggle/working/node_fidelity_complete_table_wass.csv")
```

```python
# ============================================================
# CORRECT VLM ATT (Wasserstein-weighted version)
# Concept-level ATT metrics
# Based on confidence change (conf_shift)
# ============================================================

import numpy as np
import pandas as pd
from sklearn.metrics import roc_auc_score, roc_curve, accuracy_score, f1_score

# -----------------------------
# Checks
# -----------------------------
if "node_compare_df" not in globals():
    raise ValueError("node_compare_df is missing.")

if "distance_df_wasserstein" not in globals():
    raise ValueError("distance_df_wasserstein is missing.")

# -----------------------------
# ATT score = |conf_shift|
# -----------------------------
att_df = node_compare_df.copy()
att_df["att_score"] = att_df["conf_shift"].abs()

# -----------------------------
# 🔥 CORRECT y_true (KEY FIX)
# -----------------------------
# Adaptive threshold per node (robust & concept-aware)

y_true_all = []

for node_name, g in att_df.groupby("node_name"):
    shifts = g["conf_shift"].abs().values

    if len(shifts) == 0:
        continue

    threshold = np.percentile(shifts, 75)

    for val in shifts:
        y_true_all.append(int(val >= threshold))

att_df["y_true"] = y_true_all

# -----------------------------
# Add Wasserstein weights
# -----------------------------
w_df = distance_df_wasserstein.copy().set_index("perturb_id")

if "kernel_weight_from_wassdist" not in w_df.columns:
    raise ValueError("distance_df_wasserstein must contain kernel_weight_from_wassdist")

att_df["weight"] = att_df["perturb_id"].map(
    w_df["kernel_weight_from_wassdist"]
).fillna(1.0)

# -----------------------------
# Concept mapping
# -----------------------------
def map_concept(node_name):
    s = str(node_name).lower()
    if "lesion" in s:
        return "lesion"
    elif "vessel" in s:
        return "vessel"
    elif "optic" in s:
        return "optic_disc"
    else:
        return np.nan

att_df["concept"] = att_df["node_name"].apply(map_concept)
att_df = att_df.dropna(subset=["concept"])

# -----------------------------
# Compute metrics per concept
# -----------------------------
rows = []

for concept, g in att_df.groupby("concept"):

    if len(g) < 5 or g["y_true"].nunique() < 2:
        rows.append({
            "Concept": concept,
            "ATT ACC": np.nan,
            "ATT F1": np.nan,
            "ATT AUROC": np.nan,
            "N": len(g)
        })
        continue

    y_true = g["y_true"].values
    scores = g["att_score"].values
    weights = g["weight"].values

    try:
        auc = roc_auc_score(y_true, scores, sample_weight=weights)
    except:
        auc = np.nan

    try:
        fpr, tpr, thr = roc_curve(y_true, scores)
        j = tpr - fpr
        best_thr = thr[np.argmax(j)]

        y_pred = (scores >= best_thr).astype(int)

        acc = accuracy_score(y_true, y_pred)
        f1 = f1_score(y_true, y_pred, zero_division=0)
    except:
        acc = np.nan
        f1 = np.nan

    rows.append({
        "Concept": concept,
        "ATT ACC": acc,
        "ATT F1": f1,
        "ATT AUROC": auc,
        "N": len(g)
    })

att_table_wasserstein_correct = pd.DataFrame(rows)

# -----------------------------
# Ordering
# -----------------------------
order_map = {"lesion": 0, "vessel": 1, "optic_disc": 2}
att_table_wasserstein_correct["__order"] = att_table_wasserstein_correct["Concept"].map(order_map)
att_table_wasserstein_correct = att_table_wasserstein_correct.sort_values("__order").drop(columns="__order")

print("✅ Correct Concept-level ATT Table (Wasserstein):")
display(att_table_wasserstein_correct.round(4))

# Save
att_table_wasserstein_correct.to_csv("/kaggle/working/att_table_wasserstein_correct.csv", index=False)
```

```python
# ============================================================
# FINAL VLM PATHWAY CELL:
# Surrogate and Locality Comparison Table
#
# Fills:
# ConceptLIME      = C  + WLR
# ConceptBayLIME   = C  + BRR
# ConceptSMILE-WLR = WD + WLR
# ConceptLIME-XGB  = C  + XGB
# ConceptSMILE     = WD + XGB
#
# Output table:
# Variant | Dist. | Surr. | Purpose | WMSE | WMAE | R2 | R2_w
# ============================================================

import os
import inspect
import numpy as np
import pandas as pd

from sklearn.linear_model import LinearRegression, BayesianRidge
from sklearn.metrics import mean_squared_error, mean_absolute_error, r2_score
from sklearn.model_selection import train_test_split
from xgboost import XGBRegressor


# ------------------------------------------------------------
# 1. Safety checks
# ------------------------------------------------------------

if "perturb_vectors" not in globals():
    raise ValueError("perturb_vectors is missing. Run the VLM perturbation cell first.")

X_all_full = np.asarray(perturb_vectors, dtype=float)

if X_all_full.ndim != 2:
    raise ValueError("perturb_vectors must be a 2D array.")


# ------------------------------------------------------------
# 2. Build or load VLM concept/node shift matrix
# ------------------------------------------------------------
# Preferred option:
#   node_compare_df with columns:
#   perturb_id, node_name, conf_shift
#
# Fallback option:
#   node_shift_matrix already exists

if "node_compare_df" in globals() and isinstance(node_compare_df, pd.DataFrame):

    required_cols = ["perturb_id", "node_name", "conf_shift"]
    missing_cols = [c for c in required_cols if c not in node_compare_df.columns]

    if missing_cols:
        raise ValueError(
            f"node_compare_df exists but is missing required columns: {missing_cols}"
        )

    vlm_shift_matrix = (
        node_compare_df
        .pivot_table(
            index="perturb_id",
            columns="node_name",
            values="conf_shift",
            aggfunc="first"
        )
        .sort_index()
    )

    valid_perturb_ids = vlm_shift_matrix.index.to_numpy().astype(int)

    if valid_perturb_ids.max() >= len(X_all_full):
        raise ValueError(
            "Some perturb_id values in node_compare_df exceed the length of perturb_vectors."
        )

    X_all = X_all_full[valid_perturb_ids]

    print("Using node_compare_df to build VLM concept-shift matrix.")

elif "vlm_concept_shift_matrix" in globals():

    vlm_shift_matrix = vlm_concept_shift_matrix.copy()

    if not isinstance(vlm_shift_matrix, pd.DataFrame):
        vlm_shift_matrix = pd.DataFrame(vlm_shift_matrix)

    valid_perturb_ids = np.arange(len(vlm_shift_matrix))
    X_all = X_all_full[valid_perturb_ids]

    print("Using existing vlm_concept_shift_matrix.")

elif "node_shift_matrix" in globals():

    vlm_shift_matrix = node_shift_matrix.copy()

    if not isinstance(vlm_shift_matrix, pd.DataFrame):
        vlm_shift_matrix = pd.DataFrame(vlm_shift_matrix)

    valid_perturb_ids = np.arange(len(vlm_shift_matrix))
    X_all = X_all_full[valid_perturb_ids]

    print("Using existing node_shift_matrix as the VLM shift matrix.")

else:
    raise ValueError(
        "No VLM shift matrix found. You need one of these:\n"
        "1) node_compare_df with perturb_id, node_name, conf_shift\n"
        "2) vlm_concept_shift_matrix\n"
        "3) node_shift_matrix"
    )

if vlm_shift_matrix.empty:
    raise ValueError("The VLM shift matrix is empty.")

if len(X_all) != len(vlm_shift_matrix):
    raise ValueError(
        f"Mismatch: X_all has {len(X_all)} rows but vlm_shift_matrix has "
        f"{len(vlm_shift_matrix)} rows."
    )

print("VLM shift matrix shape:", vlm_shift_matrix.shape)
print("Perturbation matrix shape:", X_all.shape)


# ------------------------------------------------------------
# 3. Metric utilities
# ------------------------------------------------------------

def weighted_mean(x, w):
    x = np.asarray(x, dtype=float)
    w = np.asarray(w, dtype=float)

    if len(x) == 0:
        return np.nan

    w_sum = np.sum(w)

    if w_sum <= 1e-12:
        return float(np.mean(x))

    return float(np.sum(w * x) / w_sum)


def weighted_mse(y_true, y_pred, w):
    return weighted_mean((np.asarray(y_true) - np.asarray(y_pred)) ** 2, w)


def weighted_mae(y_true, y_pred, w):
    return weighted_mean(np.abs(np.asarray(y_true) - np.asarray(y_pred)), w)


def weighted_r2(y_true, y_pred, w):
    y_true = np.asarray(y_true, dtype=float)
    y_pred = np.asarray(y_pred, dtype=float)
    w = np.asarray(w, dtype=float)

    if len(y_true) < 2:
        return np.nan

    y_bar = weighted_mean(y_true, w)

    ss_res = np.sum(w * (y_true - y_pred) ** 2)
    ss_tot = np.sum(w * (y_true - y_bar) ** 2)

    if ss_tot <= 1e-12:
        return np.nan

    return float(1.0 - (ss_res / ss_tot))


# ------------------------------------------------------------
# 4. Weight loaders
# ------------------------------------------------------------

def get_cosine_weight_map(n_rows):
    """
    Load cosine locality weights for the VLM/DINO pathway.
    """

    possible_vars = [
        "distance_df",
        "distance_df_cosine",
        "dino_cosine_distance_df",
        "vlm_cosine_distance_df"
    ]

    df = None
    source = None

    for var_name in possible_vars:
        if var_name in globals():
            df = globals()[var_name].copy()
            source = var_name
            break

    if df is None:
        possible_paths = [
            "/kaggle/working/dino_cosine_distance_weights.csv",
            "/kaggle/working/distance_df_from_vlm_cosine.csv",
            "/kaggle/working/distance_df_from_dino_cosine.csv",
            "/kaggle/working/distance_df.csv"
        ]

        for path in possible_paths:
            if os.path.exists(path):
                df = pd.read_csv(path)
                source = path
                break

    if df is None:
        raise FileNotFoundError(
            "Cosine locality weights not found. Run the VLM/DINO cosine distance cell first."
        )

    if "perturb_id" not in df.columns:
        raise ValueError("Cosine distance dataframe must contain a perturb_id column.")

    if "kernel_weight_from_cosdist" in df.columns:
        wcol = "kernel_weight_from_cosdist"
    elif "kernel_weight" in df.columns:
        wcol = "kernel_weight"
    else:
        raise ValueError(
            "Cosine distance dataframe must contain kernel_weight_from_cosdist or kernel_weight."
        )

    weight_map = (
        df[["perturb_id", wcol]]
        .drop_duplicates("perturb_id")
        .assign(perturb_id=lambda d: d["perturb_id"].astype(int))
        .set_index("perturb_id")[wcol]
        .astype(float)
    )

    return weight_map, wcol, source


def get_wasserstein_weight_map(n_rows):
    """
    Load Wasserstein locality weights for the VLM/DINO pathway.
    """

    possible_vars = [
        "distance_df_wasserstein",
        "distance_df_from_vlm_wasserstein",
        "distance_df_from_dino_wasserstein",
        "dino_wasserstein_distance_df",
        "vlm_wasserstein_distance_df"
    ]

    df = None
    source = None

    for var_name in possible_vars:
        if var_name in globals():
            df = globals()[var_name].copy()
            source = var_name
            break

    if df is None:
        possible_paths = [
            "/kaggle/working/dino_wasserstein_distance_weights.csv",
            "/kaggle/working/distance_df_from_vlm_wasserstein.csv",
            "/kaggle/working/distance_df_from_dino_wasserstein.csv",
            "/kaggle/working/distance_df_wasserstein.csv"
        ]

        for path in possible_paths:
            if os.path.exists(path):
                df = pd.read_csv(path)
                source = path
                break

    if df is None:
        raise FileNotFoundError(
            "Wasserstein locality weights not found. Run the VLM/DINO Wasserstein distance cell first."
        )

    if "perturb_id" not in df.columns:
        raise ValueError("Wasserstein distance dataframe must contain a perturb_id column.")

    if "kernel_weight_from_wassdist" in df.columns:
        wcol = "kernel_weight_from_wassdist"
    elif "kernel_weight" in df.columns:
        wcol = "kernel_weight"
    else:
        raise ValueError(
            "Wasserstein distance dataframe must contain kernel_weight_from_wassdist or kernel_weight."
        )

    weight_map = (
        df[["perturb_id", wcol]]
        .drop_duplicates("perturb_id")
        .assign(perturb_id=lambda d: d["perturb_id"].astype(int))
        .set_index("perturb_id")[wcol]
        .astype(float)
    )

    return weight_map, wcol, source


cos_weight_map, cos_weight_source, cos_source_name = get_cosine_weight_map(len(X_all))
wd_weight_map, wd_weight_source, wd_source_name = get_wasserstein_weight_map(len(X_all))

print("Cosine weight source:", cos_weight_source)
print("Wasserstein weight source:", wd_weight_source)


# ------------------------------------------------------------
# 5. Surrogate model factory
# ------------------------------------------------------------

def build_surrogate_model(surrogate_name, random_state):
    if surrogate_name == "WLR":
        return LinearRegression()

    if surrogate_name == "BRR":
        return BayesianRidge()

    if surrogate_name == "XGB":
        return XGBRegressor(
            objective="reg:squarederror",
            n_estimators=200,
            max_depth=3,
            learning_rate=0.05,
            subsample=0.8,
            colsample_bytree=0.8,
            reg_lambda=1.0,
            random_state=random_state,
            n_jobs=2,
            verbosity=0
        )

    raise ValueError(f"Unknown surrogate: {surrogate_name}")


def fit_with_optional_weights(model, X_train, y_train, sample_weight):
    fit_signature = inspect.signature(model.fit)

    if "sample_weight" in fit_signature.parameters:
        model.fit(X_train, y_train, sample_weight=sample_weight)
    else:
        model.fit(X_train, y_train)

    return model


# ------------------------------------------------------------
# 6. Variant definitions
# ------------------------------------------------------------

variants = [
    {
        "Variant": "ConceptLIME",
        "Dist.": "C",
        "Surr.": "WLR",
        "Purpose": "Concept-level LIME baseline"
    },
    {
        "Variant": "ConceptBayLIME",
        "Dist.": "C",
        "Surr.": "BRR",
        "Purpose": "Bayesian LIME-style baseline"
    },
    {
        "Variant": "ConceptSMILE-WLR",
        "Dist.": "WD",
        "Surr.": "WLR",
        "Purpose": "Tests Wasserstein locality with a linear surrogate"
    },
    {
        "Variant": "ConceptLIME-XGB",
        "Dist.": "C",
        "Surr.": "XGB",
        "Purpose": "Tests XGBoost under cosine locality"
    },
    {
        "Variant": "ConceptSMILE",
        "Dist.": "WD",
        "Surr.": "XGB",
        "Purpose": "Final selected ConceptSMILE configuration"
    }
]


# ------------------------------------------------------------
# 7. Run comparison
# ------------------------------------------------------------

N_REPEATS = 30
TEST_SIZE = 0.30
MIN_VALID_SAMPLES = 8

all_variant_runs = []

for variant in variants:

    variant_name = variant["Variant"]
    dist_name = variant["Dist."]
    surrogate_name = variant["Surr."]

    if dist_name == "C":
        weight_map = cos_weight_map
    elif dist_name == "WD":
        weight_map = wd_weight_map
    else:
        raise ValueError(f"Unknown distance: {dist_name}")

    for node in vlm_shift_matrix.columns:

        y_all = vlm_shift_matrix[node].values.astype(float)

        valid_mask = ~np.isnan(y_all)

        X = X_all[valid_mask]
        y = y_all[valid_mask]

        current_perturb_ids = valid_perturb_ids[valid_mask]

        missing_ids = [
            int(pid) for pid in current_perturb_ids
            if int(pid) not in weight_map.index
        ]

        if missing_ids:
            raise ValueError(
                f"Missing {dist_name} weights for perturb_id values: {missing_ids[:10]}"
            )

        w = weight_map.reindex(current_perturb_ids).values.astype(float)

        if len(y) < MIN_VALID_SAMPLES:
            continue

        if np.std(y) < 1e-8:
            continue

        for rep in range(N_REPEATS):

            rs = 42 + rep

            X_train, X_test, y_train, y_test, w_train, w_test = train_test_split(
                X,
                y,
                w,
                test_size=TEST_SIZE,
                random_state=rs
            )

            model = build_surrogate_model(surrogate_name, random_state=rs)

            model = fit_with_optional_weights(
                model=model,
                X_train=X_train,
                y_train=y_train,
                sample_weight=w_train
            )

            y_pred = model.predict(X_test)

            mse_val = mean_squared_error(y_test, y_pred)
            mae_val = mean_absolute_error(y_test, y_pred)
            wmse_val = weighted_mse(y_test, y_pred, w_test)
            wmae_val = weighted_mae(y_test, y_pred, w_test)

            try:
                r2_val = r2_score(y_test, y_pred)
            except Exception:
                r2_val = np.nan

            r2w_val = weighted_r2(y_test, y_pred, w_test)

            all_variant_runs.append({
                "Variant": variant_name,
                "Dist.": dist_name,
                "Surr.": surrogate_name,
                "Purpose": variant["Purpose"],
                "node_name": node,
                "repeat_id": rep,
                "MSE": float(mse_val),
                "MAE": float(mae_val),
                "WMSE": float(wmse_val),
                "WMAE": float(wmae_val),
                "R2": float(r2_val) if pd.notna(r2_val) else np.nan,
                "R2_w": float(r2w_val) if pd.notna(r2w_val) else np.nan
            })


vlm_surrogate_comparison_runs_df = pd.DataFrame(all_variant_runs)

if vlm_surrogate_comparison_runs_df.empty:
    raise ValueError(
        "No valid VLM surrogate comparison results were produced. "
        "Check that the VLM confidence shifts vary across perturbations."
    )


# ------------------------------------------------------------
# 8. Aggregate into exact MedSAM-style table
# ------------------------------------------------------------

vlm_surrogate_model_selection_table_df = (
    vlm_surrogate_comparison_runs_df
    .groupby(["Variant", "Dist.", "Surr.", "Purpose"], as_index=False)
    .agg({
        "WMSE": "mean",
        "WMAE": "mean",
        "R2": "mean",
        "R2_w": "mean"
    })
)

order = {
    "ConceptLIME": 0,
    "ConceptBayLIME": 1,
    "ConceptSMILE-WLR": 2,
    "ConceptLIME-XGB": 3,
    "ConceptSMILE": 4
}

vlm_surrogate_model_selection_table_df["__order"] = (
    vlm_surrogate_model_selection_table_df["Variant"].map(order)
)

vlm_surrogate_model_selection_table_df = (
    vlm_surrogate_model_selection_table_df
    .sort_values("__order")
    .drop(columns="__order")
    .reset_index(drop=True)
)

print("VLM surrogate model and locality selection table:")
display(
    vlm_surrogate_model_selection_table_df.round({
        "WMSE": 6,
        "WMAE": 6,
        "R2": 4,
        "R2_w": 4
    })
)


# ------------------------------------------------------------
# 9. Save results
# ------------------------------------------------------------

runs_path = "/kaggle/working/vlm_surrogate_comparison_runs_df.csv"
table_path = "/kaggle/working/vlm_surrogate_model_selection_table.csv"

vlm_surrogate_comparison_runs_df.to_csv(runs_path, index=False)
vlm_surrogate_model_selection_table_df.to_csv(table_path, index=False)

print("Saved:")
print(runs_path)
print(table_path)


# ------------------------------------------------------------
# 10. Optional LaTeX-ready table
# ------------------------------------------------------------

vlm_latex_table_df = vlm_surrogate_model_selection_table_df.copy()

vlm_latex_table_df["WMSE"] = vlm_latex_table_df["WMSE"].map(lambda x: f"{x:.6f}")
vlm_latex_table_df["WMAE"] = vlm_latex_table_df["WMAE"].map(lambda x: f"{x:.6f}")
vlm_latex_table_df["R2"] = vlm_latex_table_df["R2"].map(lambda x: f"{x:.4f}")
vlm_latex_table_df["R2_w"] = vlm_latex_table_df["R2_w"].map(lambda x: f"{x:.4f}")

latex_path = "/kaggle/working/vlm_surrogate_model_selection_table_latex.txt"

latex_code = vlm_latex_table_df.to_latex(
    index=False,
    escape=False,
    caption="Surrogate model and locality selection results for the VLM pathway.",
    label="tab:vlm_surrogate_locality_selection"
)

with open(latex_path, "w") as f:
    f.write(latex_code)

print("LaTeX table saved:")
print(latex_path)
```
